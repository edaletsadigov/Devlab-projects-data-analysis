## RFM Customer segmentetation

```python
s_date = data['orderdate'].max() + pd.Timedelta(days=1)
```

```python
rfm = (data.groupby('customername').agg(
            recency=('orderdate', lambda x: (s_date - x.max()).days),
            frequency=('ordernumber', 'nunique'),
            monetary=('sales', 'sum')
        ).reset_index())

rfm.head()
               
```

```python
rfm['recency_rank'] = rfm['recency'].rank(method='first')
rfm['frequency_rank'] = rfm['frequency'].rank(method='first')
rfm['monetary_rank'] = rfm['monetary'].rank(method='first')

rfm['r_score'] = pd.qcut(rfm['recency_rank'], 5, labels=[5, 4, 3, 2, 1]).astype(int)
rfm['f_score'] = pd.qcut(rfm['frequency_rank'], 5, labels=[1, 2, 3, 4, 5]).astype(int)
rfm['m_score'] = pd.qcut(rfm['monetary_rank'], 5, labels=[1, 2, 3, 4, 5]).astype(int)
```

```python
rfm
```

```python
def assign_segment(row):
    r, f = row['r_score'], row['f_score']
    if r >= 4 and f >= 4:
        return 'Champions'
    elif r >= 3 and f >= 4:
        return 'Loyal Customers'
    elif r >= 4 and  f <= 3:
        return 'Potential Loyalists'
    elif r == 5 and f == 1:
        return 'New Customers'
    elif r >= 3 and f <= 2:
        return 'Promising'
    elif r == 3 and f == 3:
        return 'Need Attention'
    elif r <= 2 and f >= 3:
        return 'At Risk'
    elif r <= 2 and f <= 2:
        return 'Hibernating'
    else:
        return 'Other'
```

```python
rfm['segment'] = rfm.apply(assign_segment,axis=1)
```

```python
segment = rfm.groupby('segment').agg(
    customer_count=('customername', 'count'),
    total_revenue=('monetary', 'sum'),
    avg_recency=('recency', 'mean'),
    avg_frequency=('frequency', 'mean')
).round(2).sort_values('total_revenue', ascending=False)
```

```python
segment
```

```python
segment['revenue_share_pct'] = (segment['total_revenue'] / rfm['monetary'].sum() * 100).round(1)
segment
```

```python


order = segment.index.tolist()

fig = make_subplots(rows=1, cols=2, subplot_titles=('Customer Count by RFM Segment', 'Total Revenue by RFM Segment'))

fig.add_trace(
    go.Bar(x=order, y=rfm['segment'].value_counts()[order], marker_color='#2E86AB', name='Customer Count'),
    row=1, col=1
)

fig.add_trace(
    go.Bar(x=order, y=segment['total_revenue'], marker_color='#F26419', name='Total Revenue'),
    row=1, col=2
)

fig.update_yaxes(title_text='Number of Customers', row=1, col=1)
fig.update_yaxes(title_text='Total Revenue (USD)', row=1, col=2)
fig.update_xaxes(tickangle=45)

fig.update_layout(showlegend=False, template='plotly_white', height=500, width=1100)
fig.show()
```

```python
fig = px.scatter(
    rfm,
    x='frequency',
    y='monetary',
    color='segment',
    title='Customer Segments: frequency vs monetary Value',
    labels={'frequency': 'frequency (unique orders)', 'monetary': 'monetary (total sales, USD)'},
    opacity=0.7,
    color_discrete_sequence=px.colors.qualitative.Set2
)

fig.update_traces(marker=dict(size=10))
fig.update_layout(template='plotly_white', legend=dict(x=1.02, y=1, xanchor='left'))
fig.show()
```

The Champions segment shows us only 23 of 92 customers (41.5%), yet contributes the majority of total revenue

- **Champions:** The most valuable segment — should be retained through loyalty programs and early-access
- **At Risk:** Before gone inactive these customers were frequent buyers — should be targeted with reactivation campaigns (like discount offers)
- **Hibernating:** Long inactive with low frequency — win-back campaigns needed or should be reassessed (high churn probability)
- **New Customers:** Only one order so far, but newly acquired — onboarding and incentives toward a second purchase are important
