## Statistical analysis

```python
def interperet(p_value:float,alpha = 0.05):
    if p_value < alpha:
        return f"SIGNIFICANT (p={p_value:.4f})"
    else:
        return f"NOT significant (p={p_value:.4f})"
```

`Total`, `cogs`, and `gross income` are perfectly correlated (r=1.0) due to the fixed 5% margin. ### Shapiro-Wilk

```python
for col in ['total','rating']:
    stat,p = st.shapiro(data_copy[col])
    print(f'{col}: W = {stat:.4f}, {interperet(p)}')
```

`Total` and `Rating` are not normally distributed `(p < 0.001)`. Therefore, non-parametric tests are used as the main results. With n=1,000, parametric tests give similar results.

### Spending: Member vs normal customers(t-test/Mann-Whitney U)

```python
member_total = data_copy.loc[data_copy['customer_type'] == 'Member', 'total']
normal_total = data_copy.loc[data_copy['customer_type'] == 'Normal', 'total']
```

```python
t_stat, t_p = st.ttest_ind(member_total,normal_total)
u_stat, u_p = st.mannwhitneyu(member_total,normal_total)
```

```python
print(f"T-test: t={t_stat:.4f}, {interperet(t_p)}")
print(f"Mann-Whitney: U={u_stat:.1f}, {interperet(u_p)}")

print(data_copy.groupby("customer_type")["total"].agg(["mean", "median", "std"]).round(2))
```

Members spend slightly more ($327.79 vs $318.12), but the difference is not statistically significant (p > 0.5). So, membership does not lead to higher spending per visit.

### Chi-Square Tests 

```python
def chi_square_test(col1, col2):
    table = pd.crosstab(data_copy[col1], data_copy[col2])
    chi2, p, dof, _ = st.chi2_contingency(table)
    print(f"{col1} x {col2}: chi2={chi2:.2f}, p={p:.4f}")
```

```python
chi_square_test("branch", "product_line")
chi_square_test("gender", "product_line")
chi_square_test("customer_type", "gender")
```

Finding: None of the three pairs show a significant association (p > 0.05):

- Branch × Product line: no meaningful branch-based preference.
- Gender × Product line: no significant gender-based preference.
- Customer type × Gender: membership is independent of gender.

The heatmap patterns are likely random variation rather than real relationships.

### Outlier detection

```python
q1 = data_copy["total"].quantile(0.25)
q3 = data_copy["total"].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
```

```python

outliers = data_copy[(data_copy["total"] < lower_bound) | (data_copy["total"] > upper_bound)]

print(f"Q1={q1:.2f}, Q3={q3:.2f}, IQR={iqr:.2f}")
print(f"Lower bound={lower_bound:.2f}, Upper bound={upper_bound:.2f}")
print(f"Outlier count: {len(outliers)} ({len(outliers) / len(data_copy) * 100:.2f}% of rows)")
```

Only 9 rows (0.90%) fall outside the IQR bounds on `Total`. This is a very low
outlier rate — no special outlier-handling (capping, removal) is needed before further analysis.
